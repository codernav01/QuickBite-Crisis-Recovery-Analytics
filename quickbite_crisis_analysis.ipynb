## Primary Key & Foreign Key Validation

### Objective

Primary and Foreign Key validation ensures that every business entity is uniquely identified and that relationships between datasets remain consistent.

Primary Keys are expected to contain unique values without duplication, while Foreign Keys should correctly reference valid records in their corresponding dimension tables.

This assessment helps identify orphan records, duplicate identifiers, and referential integrity issues that may affect joins, KPI calculations, and business analysis.

No modifications are performed during this stage. The objective is only to validate the structural integrity of the relational datasets.

# Primary Key Validation

````python
primary_keys = {
    "dim_customer": "customer_id",
    "dim_restaurant": "restaurant_id",
    "dim_menu_item": "menu_item_id",
    "dim_delivery_partner": "delivery_partner_id",
    "fact_orders": "order_id",
    "fact_delivery_performance": "order_id"
}

pk_results = []

for table_name, column in primary_keys.items():

    df = datasets[table_name]

    pk_results.append({

        "Dataset": table_name,
        "Primary Key": column,
        "Total Records": len(df),
        "Unique Values": df[column].nunique(),
        "Duplicate Keys": df[column].duplicated().sum(),
        "Null Keys": df[column].isnull().sum()

    })

pk_results = pd.DataFrame(pk_results)

pk_results
````

> **Note:** The **`fact_order_items`** and **`fact_ratings`** datasets do not contain standalone primary keys. Therefore, they are excluded from single-column primary key validation. These tables will be evaluated separately during the data quality assessment.

# Primary Key Status

````python
pk_results["Status"] = np.where(

    (pk_results["Duplicate Keys"] == 0) &
    (pk_results["Null Keys"] == 0),
    "Valid",
    "Needs Investigation"
)

pk_results
````

# Foreign Key Validation

````python
foreign_keys = [

    ("fact_orders", "customer_id", "dim_customer", "customer_id"),
    ("fact_orders", "restaurant_id", "dim_restaurant", "restaurant_id"),
    ("fact_orders", "delivery_partner_id", "dim_delivery_partner", "delivery_partner_id"),

    ("fact_order_items", "order_id", "fact_orders", "order_id"),
    ("fact_order_items", "menu_item_id", "dim_menu_item", "menu_item_id"),
    ("fact_order_items", "restaurant_id", "dim_restaurant", "restaurant_id"),

    ("fact_delivery_performance", "order_id", "fact_orders", "order_id"),

    ("fact_ratings", "order_id", "fact_orders", "order_id"),
    ("fact_ratings", "customer_id", "dim_customer", "customer_id"),
    ("fact_ratings", "restaurant_id", "dim_restaurant", "restaurant_id")

]
````

# Referential Integrity Check

````python
fk_results = []

for child_table, fk, parent_table, pk in foreign_keys:

    child = datasets[child_table]
    parent = datasets[parent_table]

    orphan_records = child[fk].notna() & (~child[fk].isin(parent[pk]))

    fk_results.append({

        "Child Table": child_table,
        "Foreign Key": fk,
        "Parent Table": parent_table,
        "Primary Key": pk,
        "Orphan Records": orphan_records.sum()

    })

fk_results = pd.DataFrame(fk_results)

fk_results
````

# Foreign Key Status

````python
fk_results["Status"] = np.where(

    fk_results["Orphan Records"] == 0,
    "Valid",
    "Needs Investigation"
)
fk_results
````

### Business Interpretation

Some customer references in the Orders dataset and order references in the Order Items dataset do not match their corresponding parent tables.

## Data Quality Scorecard

### Objective

The Data Quality Scorecard provides an overall assessment of each dataset by evaluating key quality dimensions.

Instead of reviewing individual quality issues separately, this scorecard summarizes the overall health of every dataset using standardized quality metrics.

The assessment helps identify datasets that require immediate attention before data cleaning and ensures that business analysis is performed on reliable and trustworthy data.

No data transformations are performed during this stage.

### Data Quality Metrics

The following quality dimensions are evaluated for each dataset.

- Completeness
- Duplicate-Free Rate
- Structural Integrity
- Overall Quality Score

# Generate Data Quality Scorecard

````python
pk_status_map = dict(
    zip(
        pk_results["Dataset"],
        pk_results["Status"]
    )
)

fk_status_map = (
    fk_results
    .groupby("Child Table")["Status"]
    .apply(lambda x: (x == "Valid").mean() * 100)
    .to_dict()
)

scorecard = []

for name, df in datasets.items():

    total_cells = df.shape[0] * df.shape[1]

    missing_cells = df.isnull().sum().sum()

    duplicate_rows = df.duplicated().sum()

    completeness = round(
        ((total_cells - missing_cells) / total_cells) * 100,
        2
    )

    duplicate_free_rate = round(
        ((len(df) - duplicate_rows) / len(df)) * 100,
        2
    )

    structural_scores = []

    if name in pk_status_map:
        structural_scores.append(
            100 if pk_status_map[name] == "Valid" else 0
        )

    if name in fk_status_map:
        structural_scores.append(
            fk_status_map[name]
        )

    structural_integrity = round(
        sum(structural_scores) / len(structural_scores),
        2
    ) if structural_scores else 100.00

    overall_score = round(
        (
            completeness
            + duplicate_free_rate
            + structural_integrity
        ) / 3,
        2
    )

    scorecard.append({

        "Dataset": name,

        "Completeness (%)": completeness,

        "Duplicate-Free Rate (%)": duplicate_free_rate,

        "Structural Integrity (%)": structural_integrity,

        "Overall Score (%)": overall_score

    })

scorecard = pd.DataFrame(scorecard)

scorecard
````

# Assign Quality Grade

````python
def assign_grade(score):
    if score >= 98:
        return "Excellent"
    elif score >= 95:
        return "Very Good"
    elif score >= 90:
        return "Good"
    elif score >= 80:
        return "Fair"
    else:
        return "Poor"

scorecard["Quality Grade"] = scorecard["Overall Score (%)"].apply(assign_grade)

scorecard
````

# Rank Datasets

````python
scorecard = scorecard.sort_values(
    by="Overall Score (%)",
    ascending=False
).reset_index(drop=True)

scorecard.index = scorecard.index + 1

scorecard
````

# Quality Distribution

The objective of this section is to treat missing values identified during the Missing Value Analysis using business-driven rules.

Treatment decisions are based on the role of each variable within the analytical data model. Identifier and foreign key fields are not statistically imputed because assigning artificial identifiers could create incorrect relationships and misleading business results.

Where missing information represents an incomplete business record, the record is retained in the raw dataset and handled appropriately during downstream analysis.

The treatment process is designed to improve analytical reliability while preserving the integrity of the original business data.

## 4.4 Text Standardization

### Objective

The objective of this section is to standardize text-based variables across the QuickBite datasets to ensure consistent formatting and reliable downstream analysis.

Text inconsistencies such as leading or trailing spaces, multiple internal spaces, inconsistent capitalization, and blank text values can affect grouping, filtering, reporting, dataset integration, and dashboard development.

The standardization process focuses on formatting consistency while preserving the original business meaning of the data. No business categories or legitimate text values are redefined during this stage.

## Create Text Cleaning Function

````python
def standardize_text(series):

    series = series.astype("string")

    series = series.str.strip()

    series = series.str.replace(
        r"\s+",
        " ",
        regex=True
    )

    series = series.replace(
        ["", "nan", "None", "<NA>"],
        pd.NA
    )

    return series
````

## Standardize All Text Columns

````python
for name, df in cleaned_datasets.items():

    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for column in text_columns:

        df[column] = standardize_text(df[column])
````

## 4.8 Outlier Detection

### Objective

The objective of this section is to identify statistically extreme observations within the analysis-ready QuickBite datasets before applying any outlier treatment.

Outlier detection is performed using the Interquartile Range (IQR) method. Statistical outliers are not automatically considered data errors because extreme observations may represent genuine business events, particularly during periods of operational disruption.

The assessment therefore focuses on operational variables where extreme values can materially influence performance analysis, while preserving legitimate financial and customer transaction values for business analysis.

### Outlier Detection Strategy

The IQR method is applied to selected operational variables:

- Actual delivery time
- Expected delivery time
- Delivery distance

Financial variables such as subtotal, discount, delivery fee, and total amount are not treated as statistical outliers because unusually high values may represent legitimate customer orders, promotional activity, or long-distance transactions.

Order quantity is also retained in its original form because high quantities may represent genuine bulk or business orders.

Outlier detection identifies extreme observations but does not modify the underlying data.

## Operational variables selected for outlier assessment

````python
outlier_detection_columns = {

    "fact_delivery_performance": [

        "actual_delivery_time_mins",

        "expected_delivery_time_mins",

        "distance_km"

    ]

}

outlier_detection_columns
````

### IQR-Based Outlier Identification

For each selected operational variable, the first quartile (Q1), third quartile (Q3), and interquartile range (IQR) are calculated.

Observations below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are classified as statistical outliers.

No records are removed or modified during this detection stage.

````python
outlier_detection_summary = []

for dataset_name, columns in outlier_detection_columns.items():

    df = analysis_datasets[dataset_name]

    for column in columns:
        if column not in df.columns:
            continue
        series = df[column].dropna()
        Q1 = series.quantile(0.25)
        Q3 = series.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - (1.5 * IQR)
        upper_bound = Q3 + (1.5 * IQR)
        outlier_mask = (
            (df[column] < lower_bound)
            |
            (df[column] > upper_bound)
        )
        outlier_count = int(
            outlier_mask.sum()
        )
        outlier_percentage = (
            outlier_count / len(series) * 100
            if len(series) > 0
            else 0
        )

        outlier_detection_summary.append({
            "Dataset": dataset_name,
            "Column": column,
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Records":
                outlier_count,
            "Outlier Percentage":
                round(
                    outlier_percentage,
                    2
                )

        })

outlier_detection_summary = pd.DataFrame(
    outlier_detection_summary
)

outlier_detection_summary
````

### Business Relevance Assessment

Statistical outliers identified through the IQR method are treated as candidates for review rather than automatically classified as errors.

Extreme delivery times or distances may reflect genuine operational conditions such as traffic, geographic constraints, peak demand, or crisis-period disruption.

Therefore, the next stage applies controlled treatment only to selected operational variables. Financial and customer-behavior variables remain unchanged to preserve legitimate business variation.

````python
outlier_business_summary = outlier_detection_summary[
    [
        "Dataset",
        "Column",
        "Outlier Records",
        "Outlier Percentage"
    ]
].copy()

outlier_business_summary
````

Boxplots are used to visually assess the distribution of selected operational variables and confirm the presence of extreme observations before treatment.

### Conclusion

The outlier assessment identified statistically extreme observations within selected operational variables using the IQR method.

These observations are treated as potential operational extremes rather than automatic data errors. Financial and customer-behavior variables are intentionally excluded from statistical capping to preserve legitimate business variation.

The identified operational outliers are now ready for controlled treatment in the next section.

## 4.9 Outlier Treatment

### Objective

The objective of this section is to reduce the influence of extreme operational observations while preserving legitimate QuickBite business transactions.

Outlier treatment is applied only to selected operational variables where extreme observations may disproportionately influence performance metrics.

Winsorization or automatic outlier capping is not applied to genuine operational extremes. Statistical outliers are evaluated as potential business events before deciding whether treatment is appropriate.

For QuickBite, extreme delivery-time observations may reflect genuine crisis-period operational disruption. Therefore, these observations are retained in the analytical dataset.

### Treatment Strategy

Statistical outliers identified during the detection stage are evaluated against their business context before treatment.

For QuickBite, extreme delivery-time observations may represent genuine operational disruption during the crisis period. Therefore:

- `actual_delivery_time_mins` → Retain original values
- `expected_delivery_time_mins` → Retain original values
- `distance_km` → Retain original values

No numerical capping or record deletion is applied to these operational observations.

Financial values, order quantities, discounts, delivery fees, and total amounts are also preserved in their original form.

## Create the final analytical working layer
## Original cleaned and analysis-ready datasets remain unchanged.

````python
final_datasets = {
    name: df.copy()
    for name, df in analysis_datasets.items()
}

final_datasets.keys()
````

````python
outlier_treatment_columns = {
    "fact_delivery_performance": [
        "actual_delivery_time_mins",
        "expected_delivery_time_mins",
        "distance_km"
    ]
}

outlier_treatment_columns
````

````python
outlier_treatment_summary = []

for dataset_name, columns in outlier_treatment_columns.items():

    df = final_datasets[dataset_name]

    for column in columns:

        if column not in df.columns:
            continue

        original_bounds = outlier_detection_summary[
            (outlier_detection_summary["Dataset"] == dataset_name)
            &
            (outlier_detection_summary["Column"] == column)
        ]

        if original_bounds.empty:
            continue

        outliers_before = int(
            original_bounds["Outlier Records"].iloc[0]
        )

        lower_bound = original_bounds[
            "Lower Bound"
        ].iloc[0]

        upper_bound = original_bounds[
            "Upper Bound"
        ].iloc[0]

        outlier_treatment_summary.append({

            "Dataset": dataset_name,

            "Column": column,

            "Lower Bound": lower_bound,

            "Upper Bound": upper_bound,

            "Outliers Identified": outliers_before,

            "Values Modified": 0,

            "Treatment":
                "Retained - genuine operational observations",

            "Rows Removed": 0

        })


outlier_treatment_summary = pd.DataFrame(
    outlier_treatment_summary
)

outlier_treatment_summary
````

### Treatment Verification

The selected operational variables are re-evaluated against their original IQR boundaries after the treatment decision.

Because the identified extreme observations are retained as genuine business events, the purpose of this verification is to confirm that no records or values were artificially modified.

The presence of retained outliers is therefore expected and does not represent an unresolved data-quality failure.

````python
outlier_verification = []

for dataset_name, columns in outlier_treatment_columns.items():

    df = final_datasets[dataset_name]

    for column in columns:

        if column not in df.columns:
            continue

        original_bounds = outlier_detection_summary[
            (outlier_detection_summary["Dataset"] == dataset_name)
            &
            (outlier_detection_summary["Column"] == column)
        ]

        if original_bounds.empty:
            continue

        outliers_retained = int(
            original_bounds["Outlier Records"].iloc[0]
        )

        outlier_verification.append({

            "Dataset": dataset_name,

            "Column": column,

            "Outliers Retained":
                outliers_retained,

            "Values Modified": 0,

            "Treatment Status":
                "Retained"

        })


outlier_verification = pd.DataFrame(
    outlier_verification
)

outlier_verification
````

###  Verify that record counts were preserved after treatment

````python
record_count_validation = []

for name in analysis_datasets.keys():

    before_count = len(
        analysis_datasets[name]
    )
    after_count = len(
        final_datasets[name]
    )
    record_count_validation.append({

        "Dataset": name,
        "Records Before Treatment":
            before_count,
        "Records After Treatment":
            after_count,
        "Records Removed":
            before_count - after_count,
        "Status":
            "Valid"
            if before_count == after_count
            else "Review Required"

    })


record_count_validation = pd.DataFrame(
    record_count_validation
)

record_count_validation
````

No numerical values were capped and no business transactions were deleted.

Financial values and customer-behavior variables were also preserved in their original form to maintain the integrity of downstream revenue, customer, and operational analysis.

## 5.4 Delivery Performance Features

The delivery dataset contains actual delivery time, expected delivery time, and delivery distance.

The engineered features measure:

- Delivery delay percentage

````python
delivery = feature_datasets[
    "fact_delivery_performance"
].copy()

delivery["delivery_delay_mins"] = (
    delivery["actual_delivery_time_mins"]
    -
    delivery["expected_delivery_time_mins"]
)

delivery["delivery_delay_pct"] = np.where(

    delivery["expected_delivery_time_mins"] > 0,

    (
        delivery["delivery_delay_mins"]
        /
        delivery["expected_delivery_time_mins"]
    ) * 100,

    0
)

delivery["sla_breached"] = (
    delivery["delivery_delay_mins"] > 0
).astype("int8")

delivery["on_time_delivery_flag"] = (
    delivery["delivery_delay_mins"] <= 0
).astype("int8")

delivery["delivery_efficiency_pct"] = np.where(

    delivery["actual_delivery_time_mins"] > 0,

    (
        delivery["expected_delivery_time_mins"]
        /
        delivery["actual_delivery_time_mins"]
    ) * 100,

    np.nan
)

feature_datasets[
    "fact_delivery_performance"
] = delivery

delivery_feature_summary = pd.DataFrame({

    "Feature": [
        "delivery_delay_mins",
        "delivery_delay_pct",
        "sla_breached",
        "on_time_delivery_flag",
        "delivery_efficiency_pct"
    ],

    "Business Purpose": [
        "Measures actual delay against expected delivery time",
        "Normalizes delivery delay across different SLA durations",
        "Identifies SLA breaches",
        "Identifies orders delivered within expected time",
        "Measures delivery efficiency relative to expected time"
    ]

})

delivery_feature_summary
````

## 5.5 Order & Revenue Features

Order-level financial features are created to support revenue analysis, discount analysis, cancellation analysis, and crisis-period financial comparison.

The project preserves `total_amount` as the recorded order value rather than incorrectly treating it as pure accounting revenue.

Derived financial features therefore focus on order value, discount intensity, and cancellation-related value at risk.

````python
orders = feature_datasets["fact_orders"].copy()

orders["cancelled_flag"] = (
    orders["is_cancelled"]
    .astype("string")
    .str.strip()
    .str.upper()
    .isin(["Y", "YES", "TRUE", "1"])
    .astype("int8")
)

orders["discount_pct"] = np.where(

    orders["subtotal_amount"] > 0,

    (
        orders["discount_amount"]
        /
        orders["subtotal_amount"]
    ) * 100,

    0
)

orders["discount_applied_flag"] = (
    orders["discount_amount"] > 0
).astype("int8")

orders["cancelled_order_value"] = np.where(
    orders["cancelled_flag"] == 1,
    orders["total_amount"],
    0
)

orders["order_value"] = (
    orders["total_amount"]
)

orders["gross_order_value"] = (
    orders["subtotal_amount"]
)

feature_datasets["fact_orders"] = orders

order_feature_summary = pd.DataFrame({

    "Feature": [
        "cancelled_flag",
        "discount_pct",
        "discount_applied_flag",
        "cancelled_order_value",
        "order_value",
        "gross_order_value"
    ],

    "Business Purpose": [
        "Binary cancellation indicator",
        "Measures discount intensity",
        "Identifies discounted orders",
        "Measures order value associated with cancelled orders",
        "Recorded customer order value",
        "Pre-discount order value"
    ]

})

order_feature_summary
````

## 5.6 Customer Behavioral Features

Customer-level features aggregate transactional activity to understand purchasing frequency, spending behavior, cancellation behavior, and engagement across business phases.

The features are designed to support customer segmentation, retention analysis, churn identification, and recovery prioritization.

````python
orders = feature_datasets["fact_orders"].copy()

customer_master = feature_datasets[
    "dim_customer"
][
    [
        "customer_id",
        "city",
        "acquisition_channel"
    ]
].drop_duplicates("customer_id")

customer_features = (
    customer_master
    .set_index("customer_id")
    .copy()
)

# Overall behavior

overall_customer = (
    orders
    .groupby("customer_id")
    .agg(
        total_orders=("order_id", "nunique"),
        total_spend=("order_value", "sum"),
        average_order_value=("order_value", "mean"),
        total_cancelled_orders=("cancelled_flag", "sum"),
        last_order_date=("order_timestamp", "max")
    )
)

overall_customer["cancellation_rate_pct"] = np.where(

    overall_customer["total_orders"] > 0,

    (
        overall_customer["total_cancelled_orders"]
        /
        overall_customer["total_orders"]
    ) * 100,

    0
)

# Pre-Crisis

pre_crisis = orders[
    orders["business_phase"] == "Pre-Crisis"
]

pre_customer = (
    pre_crisis
    .groupby("customer_id")
    .agg(
        pre_crisis_orders=("order_id", "nunique"),
        pre_crisis_spend=("order_value", "sum"),
        pre_crisis_avg_order_value=("order_value", "mean")
    )
)

# Crisis

crisis = orders[
    orders["business_phase"] == "Crisis"
]

crisis_customer = (
    crisis
    .groupby("customer_id")
    .agg(
        crisis_orders=("order_id", "nunique"),
        crisis_spend=("order_value", "sum"),
        crisis_avg_order_value=("order_value", "mean")
    )
)

customer_features = (
    customer_features
    .join(overall_customer)
    .join(pre_customer)
    .join(crisis_customer)
)

customer_features = customer_features.fillna({

    "total_orders": 0,
    "total_spend": 0,
    "average_order_value": 0,
    "total_cancelled_orders": 0,
    "pre_crisis_orders": 0,
    "pre_crisis_spend": 0,
    "pre_crisis_avg_order_value": 0,
    "crisis_orders": 0,
    "crisis_spend": 0,
    "crisis_avg_order_value": 0
})

customer_features["order_change_pct"] = np.where(

    customer_features["pre_crisis_orders"] > 0,

    (
        (
            customer_features["crisis_orders"]
            -
            customer_features["pre_crisis_orders"]
        )
        /
        customer_features["pre_crisis_orders"]
    ) * 100,

    np.nan
)

customer_features["spend_change_pct"] = np.where(

    customer_features["pre_crisis_spend"] > 0,

    (
        (
            customer_features["crisis_spend"]
            -
            customer_features["pre_crisis_spend"]
        )
        /
        customer_features["pre_crisis_spend"]
    ) * 100,

    np.nan
)

customer_features.reset_index(inplace=True)

customer_features
````

## 5.7 Customer Loyalty & Churn Features

Customer loyalty is defined using pre-crisis purchase frequency.

For this project, a customer with at least five Pre-Crisis orders is classified as a loyal customer.

A loyal customer is considered churned for crisis-period analysis when the customer had no orders during the Crisis phase.

This definition is intentionally transparent and directly aligned with the primary business question on loyal customer loss.

````python
customer_features["loyal_customer_flag"] = (
    customer_features["pre_crisis_orders"] >= 5
).astype("int8")

customer_features["loyal_churn_flag"] = (

    (
        customer_features["pre_crisis_orders"] >= 5
    )
    &
    (
        customer_features["crisis_orders"] == 0
    )

).astype("int8")

customer_features["active_in_crisis_flag"] = (
    customer_features["crisis_orders"] > 0
).astype("int8")

loyalty_summary = pd.DataFrame({

    "Metric": [
        "Total Customers",
        "Loyal Customers",
        "Loyal Customers Who Stopped Ordering"
    ],

    "Value": [
        len(customer_features),
        int(
            customer_features[
                "loyal_customer_flag"
            ].sum()
        ),
        int(
            customer_features[
                "loyal_churn_flag"
            ].sum()
        )
    ]

})

loyalty_summary
````

## 5.8 High-Value Customer Segmentation

High-value customers are identified using Pre-Crisis customer spend.

The top 5% of customers by Pre-Crisis spend are classified as high-value customers.

The segmentation is based on observed customer spending rather than an arbitrary monetary threshold.

This feature directly supports the business question focused on identifying high-value customers with declining engagement during the crisis.

````python
eligible_high_value = customer_features[
    customer_features["pre_crisis_orders"] > 0
]

high_value_threshold = (
    eligible_high_value[
        "pre_crisis_spend"
    ]
    .quantile(0.95)
)

customer_features[
    "high_value_customer_flag"
] = (

    customer_features[
        "pre_crisis_spend"
    ]
    >= high_value_threshold

).astype("int8")

customer_features[
    "high_value_declining_flag"
] = (

    (
        customer_features[
            "high_value_customer_flag"
        ] == 1
    )
    &
    (
        customer_features[
            "crisis_orders"
        ] < customer_features[
            "pre_crisis_orders"
        ]
    )

).astype("int8")

high_value_summary = pd.DataFrame({

    "Metric": [
        "High-Value Threshold",
        "High-Value Customers",
        "High-Value Customers With Declining Orders"
    ],

    "Value": [
        high_value_threshold,
        int(
            customer_features[
                "high_value_customer_flag"
            ].sum()
        ),
        int(
            customer_features[
                "high_value_declining_flag"
            ].sum()
        )
    ]

})

high_value_summary
````

### Customer Recovery Segment

A business-oriented recovery segment is created by combining customer loyalty, high-value status, and crisis-period engagement.

The segments are intended for prioritization rather than predictive modelling.

The classification helps management distinguish between high-priority churned customers, declining high-value customers, loyal customers, and stable customers.

````python
customer_features["customer_recovery_segment"] = np.select(

    [

        (
            (customer_features["high_value_customer_flag"] == 1)
            &
            (customer_features["loyal_churn_flag"] == 1)
        ),

        (
            (customer_features["high_value_customer_flag"] == 1)
            &
            (customer_features["high_value_declining_flag"] == 1)
        ),

        (
            (customer_features["loyal_customer_flag"] == 1)
            &
            (customer_features["loyal_churn_flag"] == 1)
        ),

        (
            customer_features["crisis_orders"] > 0
        )
        &
        (
            customer_features["order_change_pct"] < 0
        ),

        (
            customer_features["pre_crisis_orders"] == 0
        )
        &
        (
            customer_features["crisis_orders"] > 0
        )
    ],

    [

        "High-Value Loyal Churned",

        "High-Value Declining",

        "Loyal Churned",

        "Active but Declining",

        "Crisis-Only Customer"
    ],

    default="Stable / Other"
)

customer_recovery_summary = (
    customer_features[
        "customer_recovery_segment"
    ]
    .value_counts()
    .rename_axis("Recovery Segment")
    .reset_index(
        name="Customer Count"
    )
)

customer_recovery_summary
````

A Pre-Crisis minimum order threshold of 50 orders is retained as a business eligibility condition for restaurant decline analysis.

This prevents restaurants with very small historical volumes from being incorrectly ranked as major business declines.

````python
orders = feature_datasets["fact_orders"].copy()

restaurant_master = feature_datasets[
    "dim_restaurant"
][
    [
        "restaurant_id",
        "restaurant_name",
        "city",
        "cuisine_type",
        "partner_type"
    ]
].drop_duplicates("restaurant_id")

restaurant_features = (
    restaurant_master
    .set_index("restaurant_id")
    .copy()
)

restaurant_overall = (
    orders
    .groupby("restaurant_id")
    .agg(
        total_orders=("order_id", "nunique"),
        total_order_value=("order_value", "sum"),
        average_order_value=("order_value", "mean"),
        total_cancelled_orders=("cancelled_flag", "sum")
    )
)

restaurant_pre = (
    orders[
        orders["business_phase"] == "Pre-Crisis"
    ]
    .groupby("restaurant_id")
    .agg(
        pre_crisis_orders=("order_id", "nunique"),
        pre_crisis_order_value=("order_value", "sum")
    )
)

restaurant_crisis = (
    orders[
        orders["business_phase"] == "Crisis"
    ]
    .groupby("restaurant_id")
    .agg(
        crisis_orders=("order_id", "nunique"),
        crisis_order_value=("order_value", "sum")
    )
)

restaurant_features = (
    restaurant_features
    .join(restaurant_overall)
    .join(restaurant_pre)
    .join(restaurant_crisis)
)

restaurant_features = restaurant_features.fillna({

    "total_orders": 0,
    "total_order_value": 0,
    "average_order_value": 0,
    "total_cancelled_orders": 0,
    "pre_crisis_orders": 0,
    "pre_crisis_order_value": 0,
    "crisis_orders": 0,
    "crisis_order_value": 0

})

restaurant_features[
    "order_change_pct"
] = np.where(

    restaurant_features[
        "pre_crisis_orders"
    ] > 0,

    (
        (
            restaurant_features[
                "crisis_orders"
            ]
            -
            restaurant_features[
                "pre_crisis_orders"
            ]
        )
        /
        restaurant_features[
            "pre_crisis_orders"
        ]
    ) * 100,

    np.nan
)

restaurant_activity_threshold = (
    restaurant_features.loc[
        restaurant_features["pre_crisis_orders"] > 0,
        "pre_crisis_orders"
    ].quantile(0.75)
)

restaurant_features[
    "eligible_for_decline_analysis"
] = (
    restaurant_features[
        "pre_crisis_orders"
    ] >= restaurant_activity_threshold
).astype("int8")

restaurant_features.reset_index(inplace=True)

restaurant_features = restaurant_features.sort_values(
    "order_change_pct"
)

restaurant_features.head(10)
````

## 5.10 City Performance Features

City-level features are created to measure changes in order demand and order value across the Pre-Crisis and Crisis periods.

For operational business analysis, city is derived from the restaurant location associated with each order.

This creates a consistent geographic basis for comparing restaurant-side business activity across cities.

````python
restaurant_lookup = feature_datasets[
    "dim_restaurant"
][
    [
        "restaurant_id",
        "city"
    ]
].drop_duplicates("restaurant_id")

city_orders = orders.merge(
    restaurant_lookup,
    on="restaurant_id",
    how="left"
)

city_features = (
    city_orders
    .groupby(
        ["city", "business_phase"],
        dropna=False
    )
    .agg(
        orders=("order_id", "nunique"),
        order_value=("order_value", "sum"),
        cancelled_orders=("cancelled_flag", "sum")
    )
    .reset_index()
)

city_pivot = city_features.pivot(
    index="city",
    columns="business_phase",
    values=[
        "orders",
        "order_value",
        "cancelled_orders"
    ]
).fillna(0)

city_pivot.columns = [
    f"{metric}_{phase.lower().replace('-', '_')}"
    for metric, phase in city_pivot.columns
]

city_pivot = city_pivot.reset_index()

for column in [
    "orders_pre_crisis",
    "orders_crisis",
    "order_value_pre_crisis",
    "order_value_crisis",
    "cancelled_orders_pre_crisis",
    "cancelled_orders_crisis"
]:

    if column not in city_pivot.columns:
        city_pivot[column] = 0

city_pivot[
    "order_change_pct"
] = np.where(

    city_pivot["orders_pre_crisis"] > 0,

    (
        (
            city_pivot["orders_crisis"]
            -
            city_pivot["orders_pre_crisis"]
        )
        /
        city_pivot["orders_pre_crisis"]
    ) * 100,

    np.nan
)

city_pivot[
    "order_value_change_pct"
] = np.where(

    city_pivot[
        "order_value_pre_crisis"
    ] > 0,

    (
        (
            city_pivot[
                "order_value_crisis"
            ]
            -
            city_pivot[
                "order_value_pre_crisis"
            ]
        )
        /
        city_pivot[
            "order_value_pre_crisis"
        ]
    ) * 100,

    np.nan
)

city_performance_features = city_pivot.sort_values(
    "order_change_pct"
)

city_performance_features
````

## 5.11 Basket & Order Behavior Features

Order-item data is aggregated to the order level to create basket-size and purchasing-behavior features.

These features support analysis of whether customer purchasing behavior changed during the crisis, including changes in item quantity, basket size, and order composition.

````python
# ============================================================
# PART 5.11 — BASKET & ORDER BEHAVIOUR FEATURES
# ============================================================

# ------------------------------------------------------------
# 1. LOAD ORDER ITEM DATA
# ------------------------------------------------------------

order_items = feature_datasets[
    "fact_order_items"
].copy()


# ------------------------------------------------------------
# 2. CREATE ORDER-LEVEL BASKET FEATURES
# ------------------------------------------------------------

basket_features = (

    order_items
    .groupby("order_id")
    .agg(

        total_items=(
            "quantity",
            "sum"
        ),

        unique_menu_items=(
            "menu_item_id",
            "nunique"
        ),

        basket_value=(
            "line_total",
            "sum"
        ),

        item_discount_total=(
            "item_discount",
            "sum"
        ),

        average_unit_price=(
            "unit_price",
            "mean"
        )

    )
)


# ------------------------------------------------------------
# 3. CALCULATE AVERAGE ITEM VALUE
# ------------------------------------------------------------

basket_features["average_item_value"] = np.where(

    basket_features["total_items"] > 0,

    basket_features["basket_value"]
    /
    basket_features["total_items"],

    np.nan

)


# ------------------------------------------------------------
# 4. RESET INDEX
# ------------------------------------------------------------

basket_features.reset_index(
    inplace=True
)


# ------------------------------------------------------------
# 5. MERGE BASKET FEATURES INTO ORDER DATA
# ------------------------------------------------------------

orders = feature_datasets[
    "fact_orders"
].copy()


orders = orders.merge(

    basket_features,

    on="order_id",

    how="left",

    validate="one_to_one"

)


# ------------------------------------------------------------
# 6. UPDATE FEATURE DATASET
# ------------------------------------------------------------

feature_datasets[
    "fact_orders"
] = orders


# ------------------------------------------------------------
# 7. FEATURE SUMMARY
# ------------------------------------------------------------

basket_feature_summary = pd.DataFrame({

    "Feature": [

        "total_items",

        "unique_menu_items",

        "basket_value",

        "item_discount_total",

        "average_unit_price",

        "average_item_value"

    ],

    "Business Purpose": [

        "Total quantity purchased in an order",

        "Number of distinct menu items in an order",

        "Total item-level basket value",

        "Total item-level discount applied",

        "Average listed item price",

        "Average basket value per quantity purchased"

    ]

})


display(basket_feature_summary)


# ------------------------------------------------------------
# 8. FINAL FEATURE CREATION CHECK
# ------------------------------------------------------------

required_basket_columns = [

    "total_items",

    "unique_menu_items",

    "basket_value",

    "item_discount_total",

    "average_unit_price",

    "average_item_value"

]


missing_basket_columns = [

    column

    for column in required_basket_columns

    if column not in feature_datasets[
        "fact_orders"
    ].columns

]


print("=" * 80)
print("PART 5.11 — BASKET FEATURE VALIDATION")
print("=" * 80)


if len(missing_basket_columns) == 0:

    print("STATUS: PASS")

    print(
        "All basket and order behaviour features "
        "were created successfully."
    )

    print(
        f"fact_orders rows: "
        f"{len(feature_datasets['fact_orders']):,}"
    )

    print(
        f"fact_orders columns: "
        f"{feature_datasets['fact_orders'].shape[1]}"
    )

else:

    print("STATUS: REVIEW")

    print(
        "Missing features:",
        missing_basket_columns
    )

print("=" * 80)
````

## 5.12 Rating & Sentiment Features

Customer feedback is enriched with analytical rating and sentiment indicators.

Rating bands provide an interpretable classification of customer satisfaction, while sentiment categories convert the existing sentiment score into business-friendly groups.

The existing sentiment score is retained as the underlying analytical measure.

No new sentiment model is trained during this phase.

````python
ratings = feature_datasets[
    "fact_ratings"
].copy()

ratings["rating_band"] = pd.cut(

    ratings["rating"],

    bins=[0, 2, 3, 4, 5],

    labels=[
        "Poor",
        "Average",
        "Good",
        "Excellent"
    ],

    include_lowest=True
)

ratings["sentiment_category"] = np.select(

    [
        ratings["sentiment_score"] < 0,
        ratings["sentiment_score"] == 0,
        ratings["sentiment_score"] > 0
    ],

    [
        "Negative",
        "Neutral",
        "Positive"
    ],

    default="Unknown"
)

ratings["negative_review_flag"] = (
    ratings["sentiment_category"]
    == "Negative"
).astype("int8")

ratings["low_rating_flag"] = (
    ratings["rating"] <= 2
).astype("int8")

# Add business phase through the associated order

order_phase = feature_datasets[
    "fact_orders"
][
    [
        "order_id",
        "business_phase"
    ]
].drop_duplicates("order_id")

ratings = ratings.merge(
    order_phase,
    on="order_id",
    how="left"
)

feature_datasets[
    "fact_ratings"
] = ratings

rating_feature_summary = pd.DataFrame({

    "Feature": [
        "rating_band",
        "sentiment_category",
        "negative_review_flag",
        "low_rating_flag",
        "business_phase"
    ],

    "Business Purpose": [
        "Groups ratings into satisfaction bands",
        "Classifies sentiment using existing sentiment score",
        "Identifies negative feedback",
        "Identifies low customer ratings",
        "Supports Pre-Crisis vs Crisis feedback comparison"
    ]

})

rating_feature_summary
````

### Customer Cuisine Preference

Customer cuisine preference is derived from Pre-Crisis ordering behavior.

The preferred cuisine is defined as the cuisine type with the highest number of Pre-Crisis orders for each customer.

This feature helps identify whether customer recovery opportunities are concentrated within specific cuisine categories.

````python
restaurant_cuisine = feature_datasets[
    "dim_restaurant"
][
    [
        "restaurant_id",
        "cuisine_type"
    ]
].drop_duplicates("restaurant_id")

customer_cuisine_orders = (
    feature_datasets["fact_orders"]
    [
        [
            "customer_id",
            "restaurant_id",
            "business_phase"
        ]
    ]
    .merge(
        restaurant_cuisine,
        on="restaurant_id",
        how="left"
    )
)

customer_cuisine_orders = (
    customer_cuisine_orders[
        customer_cuisine_orders[
            "business_phase"
        ] == "Pre-Crisis"
    ]
)

cuisine_counts = (
    customer_cuisine_orders
    .groupby(
        [
            "customer_id",
            "cuisine_type"
        ]
    )
    .size()
    .reset_index(
        name="pre_crisis_cuisine_orders"
    )
)

preferred_cuisine = (
    cuisine_counts
    .sort_values(
        [
            "customer_id",
            "pre_crisis_cuisine_orders",
            "cuisine_type"
        ],
        ascending=[True, False, True]
    )
    .drop_duplicates("customer_id")
    [
        [
            "customer_id",
            "cuisine_type"
        ]
    ]
    .rename(
        columns={
            "cuisine_type":
                "preferred_cuisine"
        }
    )
)

customer_features = customer_features.merge(
    preferred_cuisine,
    on="customer_id",
    how="left"
)
````

## 5.13 Feature Validation

The engineered features are validated to confirm that derived values are logically consistent, business rules are preserved, and no unexpected duplicate records or key-integrity issues were introduced.

Validation focuses on:

- Feature dataset availability
- Order-level uniqueness
- Delivery delay calculation
- SLA breach logic
- Customer loyalty and churn logic
- High-value customer segmentation
- Restaurant decline-analysis eligibility

````python
feature_validation = []

# ------------------------------------------------------------
# 1. FEATURE DATASET AVAILABILITY
# ------------------------------------------------------------

required_feature_datasets = [
    "fact_orders",
    "fact_delivery_performance",
    "fact_ratings",
    "dim_customer",
    "dim_restaurant",
    "fact_order_items"
]

for name in required_feature_datasets:

    dataset_exists = (
        name in feature_datasets
        and feature_datasets[name] is not None
    )

    feature_validation.append({

        "Validation":
            f"{name} Available",

        "Invalid Records":
            0 if dataset_exists else 1,

        "Status":
            "Valid"
            if dataset_exists
            else "Review Required"

    })


# ------------------------------------------------------------
# 2. ORDER-LEVEL DUPLICATE IDs
# ------------------------------------------------------------

orders = feature_datasets["fact_orders"]

order_duplicates = int(
    orders["order_id"].duplicated().sum()
)

feature_validation.append({

    "Validation":
        "Order-Level Duplicate IDs",

    "Invalid Records":
        order_duplicates,

    "Status":
        "Valid"
        if order_duplicates == 0
        else "Review Required"

})


# ------------------------------------------------------------
# 3. DELIVERY DELAY CALCULATION
# ------------------------------------------------------------

delivery = feature_datasets[
    "fact_delivery_performance"
].copy()

expected_delay = (
    delivery["actual_delivery_time_mins"]
    -
    delivery["expected_delivery_time_mins"]
)

delivery_delay_errors = int(
    (
        ~np.isclose(
            delivery["delivery_delay_mins"],
            expected_delay,
            equal_nan=True
        )
    ).sum()
)

feature_validation.append({

    "Validation":
        "Delivery Delay Calculation",

    "Invalid Records":
        delivery_delay_errors,

    "Status":
        "Valid"
        if delivery_delay_errors == 0
        else "Review Required"

})


# ------------------------------------------------------------
# 4. SLA BREACH LOGIC
# ------------------------------------------------------------

expected_sla_flag = (
    delivery["delivery_delay_mins"] > 0
).astype("int8")

sla_logic_errors = int(
    (
        delivery["sla_breached"]
        !=
        expected_sla_flag
    ).sum()
)

feature_validation.append({

    "Validation":
        "SLA Breach Logic",

    "Invalid Records":
        sla_logic_errors,

    "Status":
        "Valid"
        if sla_logic_errors == 0
        else "Review Required"

})


# ------------------------------------------------------------
# 5. CUSTOMER LOYALTY / CHURN LOGIC
# ------------------------------------------------------------

expected_loyal_churn = (

    (
        customer_features[
            "pre_crisis_orders"
        ] >= 5
    )
    &
    (
        customer_features[
            "crisis_orders"
        ] == 0
    )

).astype("int8")

loyalty_logic_errors = int(
    (
        customer_features[
            "loyal_churn_flag"
        ]
        !=
        expected_loyal_churn
    ).sum()
)

feature_validation.append({

    "Validation":
        "Customer Loyalty / Churn Logic",

    "Invalid Records":
        loyalty_logic_errors,

    "Status":
        "Valid"
        if loyalty_logic_errors == 0
        else "Review Required"

})


# ------------------------------------------------------------
# 6. HIGH-VALUE CUSTOMER THRESHOLD
# ------------------------------------------------------------

expected_high_value = (

    customer_features[
        "pre_crisis_spend"
    ]
    >=
    high_value_threshold

).astype("int8")

high_value_errors = int(
    (
        customer_features[
            "high_value_customer_flag"
        ]
        !=
        expected_high_value
    ).sum()
)

feature_validation.append({

    "Validation":
        "High-Value Customer Threshold",

    "Invalid Records":
        high_value_errors,

    "Status":
        "Valid"
        if high_value_errors == 0
        else "Review Required"

})


# ------------------------------------------------------------
# 7. RESTAURANT DECLINE-ANALYSIS ELIGIBILITY
# ------------------------------------------------------------

restaurant_activity_threshold = (
    restaurant_features.loc[
        restaurant_features["pre_crisis_orders"] > 0,
        "pre_crisis_orders"
    ].quantile(0.75)
)

expected_restaurant_eligibility = (

    restaurant_features[
        "pre_crisis_orders"
    ]
    >=
    restaurant_activity_threshold

).astype("int8")

restaurant_eligibility_errors = int(
    (
        restaurant_features[
            "eligible_for_decline_analysis"
        ]
        !=
        expected_restaurant_eligibility
    ).sum()
)

feature_validation.append({

    "Validation":
        "Restaurant Eligibility Logic",

    "Invalid Records":
        restaurant_eligibility_errors,

    "Status":
        "Valid"
        if restaurant_eligibility_errors == 0
        else "Review Required"

})


# ------------------------------------------------------------
# FINAL VALIDATION TABLE
# ------------------------------------------------------------

feature_validation = pd.DataFrame(
    feature_validation
)

display(feature_validation)
````

## 7.3 Primary Business Question 1

### Business Question

How did customer orders change before and during the crisis?

### Business Objective

The objective is to quantify the change in order demand between the Pre-Crisis and Crisis phases and determine the magnitude of the demand contraction.

The analysis compares total orders, average monthly orders, and the percentage change between the two business phases.

## PRIMARY BUSINESS QUESTION 1

````python
q1_orders = (
    bq_orders
    .groupby("business_phase")
    .agg(
        total_orders=("order_id", "nunique"),
        total_order_value=("order_value", "sum"),
        average_order_value=("order_value", "mean")
    )
    .reset_index()
)

q1_monthly = (
    bq_orders
    .assign(
        month=bq_orders["order_timestamp"].dt.to_period("M")
    )
    .groupby("month")
    .agg(
        orders=("order_id", "nunique"),
        order_value=("order_value", "sum")
    )
    .reset_index()
)

q1_monthly["month"] = q1_monthly["month"].astype(str)

q1_pre = q1_orders[
    q1_orders["business_phase"] == "Pre-Crisis"
].iloc[0]

q1_crisis = q1_orders[
    q1_orders["business_phase"] == "Crisis"
].iloc[0]

q1_order_change_pct = (
    (
        q1_crisis["total_orders"]
        -
        q1_pre["total_orders"]
    )
    /
    q1_pre["total_orders"]
) * 100

q1_result = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Order Value",
        "Average Order Value"
    ],
    "Pre-Crisis": [
        q1_pre["total_orders"],
        q1_pre["total_order_value"],
        q1_pre["average_order_value"]
    ],
    "Crisis": [
        q1_crisis["total_orders"],
        q1_crisis["total_order_value"],
        q1_crisis["average_order_value"]
    ]
})

q1_result["Change %"] = (
    (
        q1_result["Crisis"]
        -
        q1_result["Pre-Crisis"]
    )
    /
    q1_result["Pre-Crisis"]
) * 100

display(q1_result.round(2))

print(
    f"Order volume change from Pre-Crisis to Crisis: "
    f"{q1_order_change_pct:.2f}%"
)
````

## 7.9 Primary Business Question 7

### Business Question

Which customer complaints appeared most frequently during the crisis?

### Business Objective

The objective is to identify the most common complaint themes within negative customer feedback during the Crisis phase.

The available `review_text` field is used to create transparent rule-based complaint categories.

This is a business-text classification approach rather than a machine-learning or advanced NLP model.

## PRIMARY BUSINESS QUESTION 7

````python
import re

q7_reviews = bq_ratings.copy()

q7_reviews["review_text_clean"] = (
    q7_reviews["review_text"]
    .fillna("")
    .astype(str)
    .str.lower()
)

# Negative customer feedback

q7_negative = q7_reviews[
    (
        q7_reviews["business_phase"] == "Crisis"
    )
    &
    (
        (q7_reviews["sentiment_category"] == "Negative")
        |
        (q7_reviews["low_rating_flag"] == 1)
    )
].copy()


def classify_complaint(text):

    if re.search(
        r"late|delay|delayed|slow|waiting|waited|delivery",
        text
    ):
        return "Delivery Delay"

    elif re.search(
        r"cold|taste|tasteless|quality|bad food|stale|raw|burnt|food",
        text
    ):
        return "Food Quality"

    elif re.search(
        r"wrong|missing|incorrect|item|order",
        text
    ):
        return "Order Accuracy"

    elif re.search(
        r"cancel|cancelled|refund|refunded|money back",
        text
    ):
        return "Cancellation / Refund"

    elif re.search(
        r"price|expensive|cost|charge|charged|payment|pay",
        text
    ):
        return "Price / Payment"

    elif re.search(
        r"pack|packaging|leak|spilled|spill|container",
        text
    ):
        return "Packaging"

    elif re.search(
        r"restaurant|staff|service|behavior|rude",
        text
    ):
        return "Restaurant / Service"

    else:
        return "Other Negative Feedback"


q7_negative["complaint_theme"] = (
    q7_negative["review_text_clean"]
    .apply(classify_complaint)
)


q7_theme_summary = (
    q7_negative
    .groupby("complaint_theme")
    .agg(
        complaint_count=("order_id", "count")
    )
    .reset_index()
    .sort_values(
        "complaint_count",
        ascending=False
    )
)

q7_theme_summary["share_pct"] = (
    q7_theme_summary["complaint_count"]
    /
    q7_theme_summary["complaint_count"].sum()
) * 100


display(
    q7_theme_summary.round(2)
)
````

````python
plt.figure(figsize=(11, 6))

plt.barh(
    q7_theme_summary["complaint_theme"],
    q7_theme_summary["complaint_count"]
)

plt.title("Customer Complaint Themes During the Crisis")
plt.xlabel("Number of Negative Feedback Records")
plt.ylabel("Complaint Theme")
plt.tight_layout()
plt.show()
````

### Observation

Negative customer feedback can be grouped into recurring operational and experience-related themes.

### Business Insight

The complaint distribution identifies the operational areas that generated the greatest concentration of customer dissatisfaction.

Because the classification is rule-based, the results should be interpreted as directional complaint themes rather than a formal NLP taxonomy.

### Business Recommendation

Management should prioritize remediation based on complaint frequency and business impact, particularly where complaint themes overlap with measurable operational deterioration.

### Conclusion

Customer complaint analysis provides a direct bridge between operational performance and customer experience, helping management identify the most visible sources of dissatisfaction.

## 7.13 Executive Business Question 1

### Business Question

Which operational and customer-experience factors are most strongly associated with loyal customer churn?

### Business Objective

The objective is to identify measurable customer-level risk signals associated with loyal customer churn.

The analysis evaluates cancellation behaviour, delivery SLA exposure, negative customer feedback, and exposure to declining restaurant partners.

The analysis identifies association and relative risk signals. It does not claim causal attribution because the available data is observational.

# EXECUTIVE BUSINESS QUESTION 1
## LOYAL CUSTOMER CHURN DRIVER ANALYSIS

````python
q11_orders = bq_orders[
    [
        "order_id",
        "customer_id",
        "restaurant_id",
        "business_phase",
        "cancelled_flag"
    ]
].copy()

# ------------------------------------------------------------
# CUSTOMER-LEVEL DELIVERY SLA EXPOSURE
# ------------------------------------------------------------

q11_delivery = (
    bq_delivery[
        [
            "order_id",
            "sla_breached"
        ]
    ]
    .merge(
        q11_orders[
            [
                "order_id",
                "customer_id",
                "business_phase"
            ]
        ],
        on="order_id",
        how="inner"
    )
)

q11_customer_delivery = (
    q11_delivery[
        q11_delivery[
            "business_phase"
        ] == "Pre-Crisis"
    ]
    .groupby("customer_id")
    .agg(
        pre_crisis_sla_breach_rate=(
            "sla_breached",
            "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# CUSTOMER-LEVEL NEGATIVE FEEDBACK
# ------------------------------------------------------------

q11_customer_ratings = (
    bq_ratings
    [
        [
            "customer_id",
            "negative_review_flag"
        ]
    ]
    .groupby("customer_id")
    .agg(
        negative_review_rate=(
            "negative_review_flag",
            "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# CUSTOMER EXPOSURE TO DECLINING RESTAURANTS
# ------------------------------------------------------------

q11_restaurant_change = (
    bq_restaurants[
        [
            "restaurant_id",
            "order_change_pct",
            "eligible_for_decline_analysis"
        ]
    ]
)

q11_restaurant_exposure = (
    q11_orders[
        q11_orders[
            "business_phase"
        ] == "Pre-Crisis"
    ]
    .merge(
        q11_restaurant_change,
        on="restaurant_id",
        how="left"
    )
)

q11_restaurant_exposure[
    "declining_restaurant_flag"
] = (
    (
        q11_restaurant_exposure[
            "eligible_for_decline_analysis"
        ] == 1
    )
    &
    (
        q11_restaurant_exposure[
            "order_change_pct"
        ] < 0
    )
).astype("int8")

q11_customer_restaurant = (
    q11_restaurant_exposure
    .groupby("customer_id")
    .agg(
        declining_restaurant_exposure=(
            "declining_restaurant_flag",
            "mean"
        )
    )
    .reset_index()
)

# ------------------------------------------------------------
# CUSTOMER MASTER
# ------------------------------------------------------------

q11_customer = (
    bq_customers[
        [
            "customer_id",
            "loyal_customer_flag",
            "loyal_churn_flag",
            "cancellation_rate_pct"
        ]
    ]
    .merge(
        q11_customer_delivery,
        on="customer_id",
        how="left"
    )
    .merge(
        q11_customer_ratings,
        on="customer_id",
        how="left"
    )
    .merge(
        q11_customer_restaurant,
        on="customer_id",
        how="left"
    )
)

q11_customer[
    [
        "pre_crisis_sla_breach_rate",
        "negative_review_rate",
        "declining_restaurant_exposure"
    ]
] = (
    q11_customer[
        [
            "pre_crisis_sla_breach_rate",
            "negative_review_rate",
            "declining_restaurant_exposure"
        ]
    ]
    .fillna(0)
)

q11_loyal = q11_customer[
    q11_customer[
        "loyal_customer_flag"
    ] == 1
].copy()

# ------------------------------------------------------------
# DRIVER ANALYSIS
# ------------------------------------------------------------

q11_driver_columns = {

    "High Cancellation Exposure":
        "cancellation_rate_pct",

    "High Delivery SLA Exposure":
        "pre_crisis_sla_breach_rate",

    "High Negative Feedback Exposure":
        "negative_review_rate",

    "High Declining-Restaurant Exposure":
        "declining_restaurant_exposure"

}

q11_results = []

for driver_name, column in q11_driver_columns.items():

    threshold = (
        q11_loyal[column]
        .median()
    )

    exposed = q11_loyal[
        q11_loyal[column] > threshold
    ]

    unexposed = q11_loyal[
        q11_loyal[column] <= threshold
    ]

    exposed_rate = (
        exposed[
            "loyal_churn_flag"
        ].mean()
        * 100
    )

    unexposed_rate = (
        unexposed[
            "loyal_churn_flag"
        ].mean()
        * 100
    )

    risk_gap = (
        exposed_rate
        -
        unexposed_rate
    )

    excess_churn_proxy = (
        max(risk_gap, 0)
        /
        100
        *
        len(exposed)
    )

    q11_results.append({

        "Driver":
            driver_name,

        "Median Threshold":
            threshold,

        "Exposed Customers":
            len(exposed),

        "Churn Rate - Exposed (%)":
            exposed_rate,

        "Churn Rate - Unexposed (%)":
            unexposed_rate,

        "Churn Rate Gap (pp)":
            risk_gap,

        "Excess Churn Proxy":
            excess_churn_proxy

    })

q11_driver_summary = (
    pd.DataFrame(q11_results)
    .sort_values(
        "Excess Churn Proxy",
        ascending=False
    )
    .reset_index(drop=True)
)

q11_total_proxy = (
    q11_driver_summary[
        "Excess Churn Proxy"
    ]
    .clip(lower=0)
    .sum()
)

q11_driver_summary[
    "Relative Impact Share (%)"
] = np.where(
    q11_total_proxy > 0,
    (
        q11_driver_summary[
            "Excess Churn Proxy"
        ]
        /
        q11_total_proxy
    ) * 100,
    0
)

display(
    q11_driver_summary.round(2)
)
````

````python
plt.figure(figsize=(11, 6))

plt.bar(
    q11_driver_summary["Driver"],
    q11_driver_summary[
        "Excess Churn Proxy"
    ]
)

plt.title(
    "Relative Loyal-Customer Churn Risk by Business Driver"
)

plt.xlabel("Business Driver")
plt.ylabel("Excess Churn Proxy")

plt.xticks(
    rotation=35,
    ha="right"
)

plt.tight_layout()

plt.show()
````

### Observation

The driver analysis identifies customer-level operational and experience exposures associated with different loyal-customer churn rates.

### Business Insight

The highest-ranked driver represents the strongest relative churn-risk signal among the variables evaluated.

The result is an association-based prioritization and should not be interpreted as proof of direct causality.

### Business Recommendation

### Conclusion

The analysis provides an evidence-based framework for identifying customer churn risk drivers and prioritizing management attention.

## 7.14 Executive Business Question 2

### Business Question

Which customer segments and restaurant partners should be prioritized for recovery?

### Business Objective

The objective is to rank customer recovery segments and restaurant partners according to historical business value, engagement deterioration, and recovery relevance.

The resulting score is a decision-support index and not a predictive model or financial ROI estimate.

# EXECUTIVE BUSINESS QUESTION 2
## CUSTOMER RECOVERY PRIORITIZATION

````python
# ============================================================
# EXECUTIVE BUSINESS QUESTION 2
# CUSTOMER RECOVERY PRIORITIZATION
# ============================================================

q12_customer_segments = (
    bq_customers
    .groupby(
        "customer_recovery_segment"
    )
    .agg(

        customers=(
            "customer_id",
            "nunique"
        ),

        total_pre_crisis_spend=(
            "pre_crisis_spend",
            "sum"
        ),

        average_pre_crisis_spend=(
            "pre_crisis_spend",
            "mean"
        ),

        average_pre_crisis_orders=(
            "pre_crisis_orders",
            "mean"
        ),

        average_order_change_pct=(
            "order_change_pct",
            "mean"
        ),

        churn_rate=(
            "loyal_churn_flag",
            "mean"
        ),

        active_rate=(
            "active_in_crisis_flag",
            "mean"
        )

    )
    .reset_index()
)


# ------------------------------------------------------------
# RATE CONVERSION
# ------------------------------------------------------------

q12_customer_segments[
    "churn_rate"
] *= 100

q12_customer_segments[
    "active_rate"
] *= 100


# ------------------------------------------------------------
# CUSTOMER VALUE SCORE
# ------------------------------------------------------------

max_pre_crisis_spend = (
    q12_customer_segments[
        "total_pre_crisis_spend"
    ].max()
)

q12_customer_segments[
    "value_score"
] = np.where(

    max_pre_crisis_spend > 0,

    (
        q12_customer_segments[
            "total_pre_crisis_spend"
        ]
        /
        max_pre_crisis_spend
    ) * 100,

    0

)


# ------------------------------------------------------------
# DECLINE SCORE
# ------------------------------------------------------------

q12_customer_segments[
    "decline_score"
] = (
    q12_customer_segments[
        "average_order_change_pct"
    ]
    .fillna(0)
    .clip(upper=0)
    .abs()
)


# ------------------------------------------------------------
# RECOVERY PRIORITY SCORE
# ------------------------------------------------------------

q12_customer_segments[
    "recovery_priority_score"
] = (

    q12_customer_segments[
        "value_score"
    ] * 0.40

    +

    q12_customer_segments[
        "churn_rate"
    ] * 0.30

    +

    q12_customer_segments[
        "decline_score"
    ] * 0.20

    +

    (
        100
        -
        q12_customer_segments[
            "active_rate"
        ]
    ) * 0.10

)


# ------------------------------------------------------------
# FINAL PRIORITIZATION
# ------------------------------------------------------------

q12_customer_segments = (
    q12_customer_segments
    .sort_values(
        "recovery_priority_score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


display(
    q12_customer_segments.round(2)
)
````

````python
q12_restaurants = (
    bq_restaurants[
        bq_restaurants[
            "eligible_for_decline_analysis"
        ] == 1
    ]
    .copy()
)

q12_restaurants[
    "decline_severity"
] = (
    q12_restaurants[
        "order_change_pct"
    ]
    .clip(upper=0)
    .abs()
)

q12_restaurants[
    "historical_value_score"
] = np.where(
    q12_restaurants[
        "pre_crisis_order_value"
    ].max() > 0,
    (
        q12_restaurants[
            "pre_crisis_order_value"
        ]
        /
        q12_restaurants[
            "pre_crisis_order_value"
        ].max()
    ) * 100,
    0
)

q12_restaurants[
    "engagement_score"
] = np.where(
    q12_restaurants[
        "pre_crisis_orders"
    ] > 0,
    (
        q12_restaurants[
            "crisis_orders"
        ]
        /
        q12_restaurants[
            "pre_crisis_orders"
        ]
    ).clip(0, 1) * 100,
    0
)

q12_restaurants[
    "recovery_priority_score"
] = (
    q12_restaurants[
        "historical_value_score"
    ] * 0.45
    +
    q12_restaurants[
        "decline_severity"
    ] * 0.35
    +
    (
        100
        -
        q12_restaurants[
            "engagement_score"
        ]
    ) * 0.20
)

q12_restaurant_priority = (
    q12_restaurants[
        [
            "restaurant_id",
            "restaurant_name",
            "city",
            "cuisine_type",
            "partner_type",
            "pre_crisis_orders",
            "crisis_orders",
            "pre_crisis_order_value",
            "order_change_pct",
            "recovery_priority_score"
        ]
    ]
    .sort_values(
        "recovery_priority_score",
        ascending=False
    )
)

display(
    q12_restaurant_priority
    .head(20)
    .round(2)
)
````

## 7.15 Executive Business Question 3

### Business Question

Which operational and customer-experience KPIs can serve as early-warning indicators of deteriorating business performance?

### Business Objective

The objective is to evaluate monthly operational and customer-experience indicators against subsequent business outcomes.

The analysis uses a one-month lead comparison.

Because the project contains a limited number of monthly observations, the results are directional early-warning signals rather than statistically conclusive causal relationships.

# EXECUTIVE BUSINESS QUESTION 3
## MONTHLY EARLY-WARNING ANALYSIS

### MONTHLY ORDER METRICS

````python
q13_orders = bq_orders.copy()

q13_orders["month"] = (
    q13_orders["order_timestamp"]
    .dt.to_period("M")
)

q13_monthly_orders = (
    q13_orders
    .groupby("month")
    .agg(

        orders=(
            "order_id",
            "nunique"
        ),

        order_value=(
            "order_value",
            "sum"
        ),

        cancellation_rate=(
            "cancelled_flag",
            "mean"
        ),

        active_customers=(
            "customer_id",
            "nunique"
        )

    )
)

# ------------------------------------------------------------
# DELIVERY MONTH
# IMPORTANT:
# Merge through order_id rather than assigning values
# positionally.
# ------------------------------------------------------------

q13_delivery_dates = (
    bq_orders[
        [
            "order_id",
            "order_timestamp"
        ]
    ]
    .drop_duplicates("order_id")
)

q13_delivery = (
    bq_delivery[
        [
            "order_id",
            "actual_delivery_time_mins",
            "sla_breached"
        ]
    ]
    .merge(
        q13_delivery_dates,
        on="order_id",
        how="left",
        validate="many_to_one"
    )
)

q13_delivery["month"] = (
    q13_delivery["order_timestamp"]
    .dt.to_period("M")
)

q13_monthly_delivery = (
    q13_delivery
    .groupby("month")
    .agg(

        avg_delivery_time=(
            "actual_delivery_time_mins",
            "mean"
        ),

        sla_breach_rate=(
            "sla_breached",
            "mean"
        )

    )
)

# ------------------------------------------------------------
# MONTHLY RATING METRICS
# ------------------------------------------------------------

q13_ratings = bq_ratings.copy()

q13_ratings["month"] = (
    q13_ratings["review_timestamp"]
    .dt.to_period("M")
)

q13_monthly_rating = (
    q13_ratings
    .groupby("month")
    .agg(

        average_rating=(
            "rating",
            "mean"
        ),

        negative_review_rate=(
            "negative_review_flag",
            "mean"
        )

    )
)

# ------------------------------------------------------------
# COMBINED MONTHLY KPI TABLE
# ------------------------------------------------------------

q13_monthly = (
    q13_monthly_orders
    .join(
        q13_monthly_delivery,
        how="outer"
    )
    .join(
        q13_monthly_rating,
        how="outer"
    )
    .sort_index()
    .reset_index()
)

q13_monthly[
    "cancellation_rate"
] *= 100

q13_monthly[
    "sla_breach_rate"
] *= 100

q13_monthly[
    "negative_review_rate"
] *= 100

display(
    q13_monthly.round(2)
)
````

## ONE-MONTH LEAD RELATIONSHIP

````python
q13_analysis = q13_monthly.copy()

q13_analysis[
    "next_month_order_value"
] = (
    q13_analysis[
        "order_value"
    ].shift(-1)
)

q13_analysis[
    "next_month_active_customers"
] = (
    q13_analysis[
        "active_customers"
    ].shift(-1)
)

q13_analysis[
    "next_month_rating"
] = (
    q13_analysis[
        "average_rating"
    ].shift(-1)
)

q13_results = []

q13_leading_kpis = [
    "orders",
    "cancellation_rate",
    "avg_delivery_time",
    "sla_breach_rate",
    "negative_review_rate",
    "active_customers"
]

q13_outcomes = [
    "next_month_order_value",
    "next_month_active_customers",
    "next_month_rating"
]

for kpi in q13_leading_kpis:

    for outcome in q13_outcomes:

        temp = q13_analysis[
            [
                kpi,
                outcome
            ]
        ].dropna()

        if len(temp) >= 3:

            correlation = (
                temp[kpi]
                .corr(temp[outcome])
            )

        else:

            correlation = np.nan

        q13_results.append({

            "Leading KPI":
                kpi,

            "Next-Month Outcome":
                outcome,

            "Observations":
                len(temp),

            "Correlation":
                correlation

        })

q13_lead_metrics = (
    pd.DataFrame(q13_results)
    .sort_values(
        "Correlation",
        ascending=True
    )
)

display(
    q13_lead_metrics.round(3)
)
````

## 7.16 Executive Business Question 4

### Business Question

Which recovery initiatives should receive priority when management resources are constrained?

### Business Objective

The objective is to rank major recovery initiatives according to measurable business exposure.

The project does not contain intervention cost or actual ROI data. Therefore, the analysis uses business-impact proxies rather than claiming financial return on investment.

# EXECUTIVE BUSINESS QUESTION 4
## RECOVERY RESOURCE PRIORITIZATION

````python
q14_initiatives = []


# ------------------------------------------------------------
# 1. HIGH-VALUE CUSTOMER RECOVERY
# ------------------------------------------------------------

q14_high_value = bq_customers[
    bq_customers[
        "high_value_declining_flag"
    ] == 1
].copy()

q14_customer_proxy = (
    q14_high_value[
        "pre_crisis_spend"
    ].sum()
)

q14_initiatives.append({

    "Initiative":
        "High-Value Customer Recovery",

    "Affected Population":
        len(q14_high_value),

    "Impact Proxy":
        q14_customer_proxy

})


# ------------------------------------------------------------
# 2. RESTAURANT PARTNER RECOVERY
# ------------------------------------------------------------

q14_declining_restaurants = (
    bq_restaurants[
        (
            bq_restaurants[
                "eligible_for_decline_analysis"
            ] == 1
        )
        &
        (
            bq_restaurants[
                "order_change_pct"
            ] < 0
        )
    ]
    .copy()
)

q14_restaurant_proxy = (
    q14_declining_restaurants[
        "pre_crisis_order_value"
    ].sum()
)

q14_initiatives.append({

    "Initiative":
        "Restaurant Partner Recovery",

    "Affected Population":
        len(q14_declining_restaurants),

    "Impact Proxy":
        q14_restaurant_proxy

})


# ------------------------------------------------------------
# 3. DELIVERY SLA IMPROVEMENT
# ------------------------------------------------------------

q14_delivery_phase = (
    bq_delivery
    .groupby(
        "business_phase"
    )
    .agg(

        orders=(
            "order_id",
            "nunique"
        ),

        sla_breach_rate=(
            "sla_breached",
            "mean"
        )

    )
)

q14_sla_gap = (
    q14_delivery_phase.loc[
        "Crisis",
        "sla_breach_rate"
    ]
    -
    q14_delivery_phase.loc[
        "Pre-Crisis",
        "sla_breach_rate"
    ]
)

q14_delivery_proxy = (
    max(
        q14_sla_gap,
        0
    )
    *
    q14_delivery_phase.loc[
        "Crisis",
        "orders"
    ]
)

q14_initiatives.append({

    "Initiative":
        "Delivery SLA Improvement",

    "Affected Population":
        int(
            q14_delivery_phase.loc[
                "Crisis",
                "orders"
            ]
        ),

    "Impact Proxy":
        q14_delivery_proxy

})


# ------------------------------------------------------------
# 4. CANCELLATION REDUCTION
# ------------------------------------------------------------

q14_cancellation = (
    bq_orders
    .groupby(
        "business_phase"
    )
    .agg(

        orders=(
            "order_id",
            "nunique"
        ),

        cancellation_rate=(
            "cancelled_flag",
            "mean"
        )

    )
)

q14_cancel_gap = (
    q14_cancellation.loc[
        "Crisis",
        "cancellation_rate"
    ]
    -
    q14_cancellation.loc[
        "Pre-Crisis",
        "cancellation_rate"
    ]
)

q14_cancel_proxy = (
    max(
        q14_cancel_gap,
        0
    )
    *
    q14_cancellation.loc[
        "Crisis",
        "orders"
    ]
)

q14_initiatives.append({

    "Initiative":
        "Cancellation Reduction",

    "Affected Population":
        int(
            q14_cancellation.loc[
                "Crisis",
                "orders"
            ]
        ),

    "Impact Proxy":
        q14_cancel_proxy

})


# ------------------------------------------------------------
# 5. CUSTOMER COMPLAINT REMEDIATION
# ------------------------------------------------------------

q14_negative_reviews = int(
    (
        bq_ratings[
            "negative_review_flag"
        ] == 1
    ).sum()
)

q14_initiatives.append({

    "Initiative":
        "Customer Complaint Remediation",

    "Affected Population":
        q14_negative_reviews,

    "Impact Proxy":
        q14_negative_reviews

})


# ------------------------------------------------------------
# CREATE INITIATIVE DATAFRAME
# ------------------------------------------------------------

q14_initiatives = pd.DataFrame(
    q14_initiatives
)


# ------------------------------------------------------------
# RELATIVE PRIORITY SCORE
# ------------------------------------------------------------

q14_initiatives[
    "Relative Priority Score"
] = (
    q14_initiatives[
        "Impact Proxy"
    ]
    .rank(
        method="min",
        ascending=False
    )
)


# ------------------------------------------------------------
# NORMALIZE PRIORITY SCORE TO 0–100
# ------------------------------------------------------------

q14_max_rank = (
    q14_initiatives[
        "Relative Priority Score"
    ].max()
)

if q14_max_rank > 0:

    q14_initiatives[
        "Relative Priority Score"
    ] = (
        (
            q14_max_rank
            -
            q14_initiatives[
                "Relative Priority Score"
            ]
            + 1
        )
        /
        q14_max_rank
    ) * 100

else:

    q14_initiatives[
        "Relative Priority Score"
    ] = 0


# ------------------------------------------------------------
# FINAL PRIORITIZATION
# ------------------------------------------------------------

q14_initiatives = (
    q14_initiatives
    .sort_values(
        "Relative Priority Score",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


# ------------------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------------------

display(
    q14_initiatives.round(2)
)
````

````python
plt.figure(figsize=(11, 6))

q14_plot = (
    q14_initiatives
    .sort_values(
        "Relative Priority Score"
    )
)

plt.barh(
    q14_plot[
        "Initiative"
    ],
    q14_plot[
        "Relative Priority Score"
    ]
)

plt.title(
    "Recovery Initiative Priority"
)

plt.xlabel(
    "Relative Priority Score"
)

plt.ylabel(
    "Recovery Initiative"
)

plt.tight_layout()

plt.show()
````

### Observation

The impact-proxy framework identifies recovery initiatives associated with the largest measurable business exposure.

### Business Insight

Customer recovery, restaurant partner recovery, delivery reliability, and cancellation reduction address different dimensions of the business disruption.

### Business Recommendation

Management should allocate scarce resources toward initiatives with the highest measurable business exposure while using operational improvements to protect customer experience across the platform.

### Conclusion

The prioritization framework provides a transparent decision-support mechanism without making unsupported ROI claims.

````python
# ------------------------------------------------------------
# 1. KPI MONITORING FRAMEWORK
# ------------------------------------------------------------

q15_framework = pd.DataFrame({

    "KPI": [
        "Order Volume",
        "Order Value",
        "Active Customers",
        "Cancellation Rate",
        "Average Delivery Time",
        "SLA Breach Rate",
        "Average Customer Rating",
        "Negative Review Rate",
        "Declining Restaurant Exposure"
    ],

    "Business Area": [
        "Demand",
        "Commercial",
        "Customer",
        "Operations",
        "Delivery",
        "Delivery",
        "Customer Experience",
        "Customer Experience",
        "Partner"
    ],

    "Monitoring Frequency": [
        "Daily",
        "Daily",
        "Daily",
        "Daily",
        "Daily",
        "Daily",
        "Daily",
        "Daily",
        "Weekly"
    ],

    "Warning Rule": [
        "Decline > 10%",
        "Decline > 10%",
        "Decline > 10%",
        "Increase > 10%",
        "Increase > 10%",
        "Increase > 10%",
        "Decline > 5%",
        "Increase > 10%",
        "Increase > 10%"
    ],

    "Critical Rule": [
        "Decline > 20%",
        "Decline > 20%",
        "Decline > 20%",
        "Increase > 20%",
        "Increase > 20%",
        "Increase > 20%",
        "Decline > 10%",
        "Increase > 20%",
        "Increase > 20%"
    ],

    "Executive Action": [
        "Investigate demand drivers",
        "Assess commercial exposure",
        "Launch retention intervention",
        "Investigate fulfillment failures",
        "Review delivery capacity",
        "Escalate SLA deterioration",
        "Investigate customer experience",
        "Review complaint drivers",
        "Review partner health"
    ]
})


# ------------------------------------------------------------
# 2. CALCULATE EXECUTIVE KPI STATUS
# ------------------------------------------------------------

q15_status = []


# ------------------------------------------------------------
# ORDER VOLUME
# ------------------------------------------------------------

if "bq_orders" in globals():

    order_volume = (
        bq_orders
        .groupby("business_phase")["order_id"]
        .nunique()
    )

    if (
        "Pre-Crisis" in order_volume.index
        and
        "Crisis" in order_volume.index
        and
        order_volume["Pre-Crisis"] > 0
    ):

        order_volume_change = (
            (
                order_volume["Crisis"]
                -
                order_volume["Pre-Crisis"]
            )
            /
            order_volume["Pre-Crisis"]
        ) * 100

        q15_status.append({

            "KPI": "Order Volume",

            "Change (%)": order_volume_change,

            "Status": (
                "Critical"
                if order_volume_change < -20
                else
                "Warning"
                if order_volume_change < -10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# ORDER VALUE
# ------------------------------------------------------------

if (
    "bq_orders" in globals()
    and
    "order_value" in bq_orders.columns
):

    order_value = (
        bq_orders
        .groupby("business_phase")["order_value"]
        .sum()
    )

    if (
        "Pre-Crisis" in order_value.index
        and
        "Crisis" in order_value.index
        and
        order_value["Pre-Crisis"] > 0
    ):

        order_value_change = (
            (
                order_value["Crisis"]
                -
                order_value["Pre-Crisis"]
            )
            /
            order_value["Pre-Crisis"]
        ) * 100

        q15_status.append({

            "KPI": "Order Value",

            "Change (%)": order_value_change,

            "Status": (
                "Critical"
                if order_value_change < -20
                else
                "Warning"
                if order_value_change < -10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# ACTIVE CUSTOMERS
# ------------------------------------------------------------

if "bq_customers" in globals():

    if (
        "active_in_crisis_flag" in bq_customers.columns
        and
        "pre_crisis_orders" in bq_customers.columns
    ):

        pre_crisis_active = (
            bq_customers["pre_crisis_orders"] > 0
        ).sum()

        crisis_active = (
            bq_customers["active_in_crisis_flag"] == 1
        ).sum()

        if pre_crisis_active > 0:

            active_customer_change = (
                (
                    crisis_active
                    -
                    pre_crisis_active
                )
                /
                pre_crisis_active
            ) * 100

            q15_status.append({

                "KPI": "Active Customers",

                "Change (%)": active_customer_change,

                "Status": (
                    "Critical"
                    if active_customer_change < -20
                    else
                    "Warning"
                    if active_customer_change < -10
                    else
                    "Normal"
                )
            })


# ------------------------------------------------------------
# CANCELLATION RATE
# ------------------------------------------------------------

if "bq_orders" in globals():

    cancellation = (
        bq_orders
        .groupby("business_phase")["cancelled_flag"]
        .mean()
        * 100
    )

    if (
        "Pre-Crisis" in cancellation.index
        and
        "Crisis" in cancellation.index
    ):

        cancellation_change = (
            cancellation["Crisis"]
            -
            cancellation["Pre-Crisis"]
        )

        q15_status.append({

            "KPI": "Cancellation Rate",

            "Change (%)": cancellation_change,

            "Status": (
                "Critical"
                if cancellation_change > 20
                else
                "Warning"
                if cancellation_change > 10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# AVERAGE DELIVERY TIME
# ------------------------------------------------------------

if "bq_delivery" in globals():

    delivery_time = (
        bq_delivery
        .groupby("business_phase")["actual_delivery_time_mins"]
        .mean()
    )

    if (
        "Pre-Crisis" in delivery_time.index
        and
        "Crisis" in delivery_time.index
        and
        delivery_time["Pre-Crisis"] > 0
    ):

        delivery_time_change = (
            (
                delivery_time["Crisis"]
                -
                delivery_time["Pre-Crisis"]
            )
            /
            delivery_time["Pre-Crisis"]
        ) * 100

        q15_status.append({

            "KPI": "Average Delivery Time",

            "Change (%)": delivery_time_change,

            "Status": (
                "Critical"
                if delivery_time_change > 20
                else
                "Warning"
                if delivery_time_change > 10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# SLA BREACH RATE
# ------------------------------------------------------------

if "bq_delivery" in globals():

    sla_rate = (
        bq_delivery
        .groupby("business_phase")["sla_breached"]
        .mean()
        * 100
    )

    if (
        "Pre-Crisis" in sla_rate.index
        and
        "Crisis" in sla_rate.index
    ):

        sla_change = (
            sla_rate["Crisis"]
            -
            sla_rate["Pre-Crisis"]
        )

        q15_status.append({

            "KPI": "SLA Breach Rate",

            "Change (%)": sla_change,

            "Status": (
                "Critical"
                if sla_change > 20
                else
                "Warning"
                if sla_change > 10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# AVERAGE CUSTOMER RATING
# ------------------------------------------------------------

if "bq_ratings" in globals():

    rating = (
        bq_ratings
        .groupby("business_phase")["rating"]
        .mean()
    )

    if (
        "Pre-Crisis" in rating.index
        and
        "Crisis" in rating.index
        and
        rating["Pre-Crisis"] > 0
    ):

        rating_change = (
            (
                rating["Crisis"]
                -
                rating["Pre-Crisis"]
            )
            /
            rating["Pre-Crisis"]
        ) * 100

        q15_status.append({

            "KPI": "Average Customer Rating",

            "Change (%)": rating_change,

            "Status": (
                "Critical"
                if rating_change < -10
                else
                "Warning"
                if rating_change < -5
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# NEGATIVE REVIEW RATE
# ------------------------------------------------------------

if "bq_ratings" in globals():

    negative_review_rate = (
        bq_ratings
        .groupby("business_phase")["negative_review_flag"]
        .mean()
        * 100
    )

    if (
        "Pre-Crisis" in negative_review_rate.index
        and
        "Crisis" in negative_review_rate.index
    ):

        negative_review_change = (
            negative_review_rate["Crisis"]
            -
            negative_review_rate["Pre-Crisis"]
        )

        q15_status.append({

            "KPI": "Negative Review Rate",

            "Change (%)": negative_review_change,

            "Status": (
                "Critical"
                if negative_review_change > 20
                else
                "Warning"
                if negative_review_change > 10
                else
                "Normal"
            )
        })


# ------------------------------------------------------------
# DECLINING RESTAURANT EXPOSURE
# ------------------------------------------------------------

if "restaurant_features" in globals():

    if (
        "eligible_for_decline_analysis"
        in restaurant_features.columns
        and
        "order_change_pct"
        in restaurant_features.columns
    ):

        eligible_restaurants = (
            restaurant_features[
                restaurant_features[
                    "eligible_for_decline_analysis"
                ] == 1
            ]
        )

        if len(eligible_restaurants) > 0:

            declining_restaurants = (
                eligible_restaurants[
                    eligible_restaurants[
                        "order_change_pct"
                    ] < 0
                ]
            )

            restaurant_exposure_rate = (
                len(declining_restaurants)
                /
                len(eligible_restaurants)
            ) * 100

            q15_status.append({

                "KPI": "Declining Restaurant Exposure",

                "Change (%)": restaurant_exposure_rate,

                "Status": (
                    "Critical"
                    if restaurant_exposure_rate > 20
                    else
                    "Warning"
                    if restaurant_exposure_rate > 10
                    else
                    "Normal"
                )
            })


# ------------------------------------------------------------
# FINAL STATUS TABLE
# ------------------------------------------------------------

q15_status = pd.DataFrame(q15_status)


# ------------------------------------------------------------
# DISPLAY OUTPUTS
# ------------------------------------------------------------

display(q15_framework)

display(q15_status.round(2))
````

## PRE-CRISIS VS CRISIS KPI STATUS

````python
q15_orders = (
    bq_orders
    .groupby("business_phase")
    .agg(

        orders=(
            "order_id",
            "nunique"
        ),

        order_value=(
            "order_value",
            "sum"
        ),

        cancellation_rate=(
            "cancelled_flag",
            "mean"
        )

    )
)

q15_delivery = (
    bq_delivery
    .groupby("business_phase")
    .agg(

        average_delivery_time=(
            "actual_delivery_time_mins",
            "mean"
        ),

        sla_breach_rate=(
            "sla_breached",
            "mean"
        )

    )
)

q15_rating = (
    bq_ratings
    .groupby("business_phase")
    .agg(

        average_rating=(
            "rating",
            "mean"
        ),

        negative_review_rate=(
            "negative_review_flag",
            "mean"
        )

    )
)

q15_active_customers = (
    bq_orders
    .groupby("business_phase")
    .agg(
        active_customers=(
            "customer_id",
            "nunique"
        )
    )
)

q15_orders[
    "cancellation_rate"
] *= 100

q15_delivery[
    "sla_breach_rate"
] *= 100

q15_rating[
    "negative_review_rate"
] *= 100


def q15_pct_change(
    pre_value,
    crisis_value
):

    if pd.isna(pre_value) or pd.isna(crisis_value):

        return np.nan

    if pre_value == 0:

        if crisis_value == 0:
            return 0.0

        return np.inf

    return (
        (
            crisis_value
            -
            pre_value
        )
        /
        pre_value
    ) * 100


q15_status = []

# ------------------------------------------------------------
# ORDER VOLUME
# ------------------------------------------------------------

pre_value = q15_orders.loc[
    "Pre-Crisis",
    "orders"
]

crisis_value = q15_orders.loc[
    "Crisis",
    "orders"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Order Volume",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Lower is adverse",

    "Status":
        "Critical"
        if change <= -20
        else "Warning"
        if change <= -10
        else "Stable"

})

# ------------------------------------------------------------
# ORDER VALUE
# ------------------------------------------------------------

pre_value = q15_orders.loc[
    "Pre-Crisis",
    "order_value"
]

crisis_value = q15_orders.loc[
    "Crisis",
    "order_value"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Order Value",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Lower is adverse",

    "Status":
        "Critical"
        if change <= -20
        else "Warning"
        if change <= -10
        else "Stable"

})

# ------------------------------------------------------------
# ACTIVE CUSTOMERS
# ------------------------------------------------------------

pre_value = q15_active_customers.loc[
    "Pre-Crisis",
    "active_customers"
]

crisis_value = q15_active_customers.loc[
    "Crisis",
    "active_customers"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Active Customers",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Lower is adverse",

    "Status":
        "Critical"
        if change <= -20
        else "Warning"
        if change <= -10
        else "Stable"

})

# ------------------------------------------------------------
# CANCELLATION RATE
# ------------------------------------------------------------

pre_value = q15_orders.loc[
    "Pre-Crisis",
    "cancellation_rate"
]

crisis_value = q15_orders.loc[
    "Crisis",
    "cancellation_rate"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Cancellation Rate",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Higher is adverse",

    "Status":
        "Critical"
        if change >= 20
        else "Warning"
        if change >= 10
        else "Stable"

})

# ------------------------------------------------------------
# DELIVERY TIME
# ------------------------------------------------------------

pre_value = q15_delivery.loc[
    "Pre-Crisis",
    "average_delivery_time"
]

crisis_value = q15_delivery.loc[
    "Crisis",
    "average_delivery_time"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Average Delivery Time",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Higher is adverse",

    "Status":
        "Critical"
        if change >= 20
        else "Warning"
        if change >= 10
        else "Stable"

})

# ------------------------------------------------------------
# SLA BREACH RATE
# ------------------------------------------------------------

pre_value = q15_delivery.loc[
    "Pre-Crisis",
    "sla_breach_rate"
]

crisis_value = q15_delivery.loc[
    "Crisis",
    "sla_breach_rate"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "SLA Breach Rate",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Higher is adverse",

    "Status":
        "Critical"
        if change >= 20
        else "Warning"
        if change >= 10
        else "Stable"

})

# ------------------------------------------------------------
# CUSTOMER RATING
# ------------------------------------------------------------

pre_value = q15_rating.loc[
    "Pre-Crisis",
    "average_rating"
]

crisis_value = q15_rating.loc[
    "Crisis",
    "average_rating"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Average Customer Rating",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Lower is adverse",

    "Status":
        "Critical"
        if change <= -10
        else "Warning"
        if change <= -5
        else "Stable"

})

# ------------------------------------------------------------
# NEGATIVE REVIEW RATE
# ------------------------------------------------------------

pre_value = q15_rating.loc[
    "Pre-Crisis",
    "negative_review_rate"
]

crisis_value = q15_rating.loc[
    "Crisis",
    "negative_review_rate"
]

change = q15_pct_change(
    pre_value,
    crisis_value
)

q15_status.append({

    "KPI": "Negative Review Rate",

    "Pre-Crisis": pre_value,

    "Crisis": crisis_value,

    "Change (%)": change,

    "Direction": "Higher is adverse",

    "Status":
        "Critical"
        if change >= 20
        else "Warning"
        if change >= 10
        else "Stable"

})

q15_status = pd.DataFrame(
    q15_status
)

display(
    q15_status.round(2)
)
````

````python
export_validation = []


# ------------------------------------------------------------
# CLEANED DATASETS
# ------------------------------------------------------------

for dataset_name in cleaned_datasets:

    file_path = os.path.join(

        cleaned_export_path,

        f"{dataset_name}.csv"

    )

    export_validation.append({

        "Export Group":
            "Cleaned Dataset",

        "Object":
            dataset_name,

        "File Exists":
            os.path.exists(file_path),

        "Rows":
            len(cleaned_datasets[dataset_name])

    })


# ------------------------------------------------------------
# SQL DATASETS
# ------------------------------------------------------------

for dataset_name in analysis_datasets:

    file_path = os.path.join(

        analysis_export_path,

        f"{dataset_name}.csv"

    )

    export_validation.append({

        "Export Group":
            "SQL Dataset",

        "Object":
            dataset_name,

        "File Exists":
            os.path.exists(file_path),

        "Rows":
            len(analysis_datasets[dataset_name])

    })


# ------------------------------------------------------------
# POWER BI DATASETS
# ------------------------------------------------------------

for dataset_name in powerbi_datasets:

    file_path = os.path.join(

        powerbi_export_path,

        f"{dataset_name}.csv"

    )

    export_validation.append({

        "Export Group":
            "Power BI Dataset",

        "Object":
            dataset_name,

        "File Exists":
            os.path.exists(file_path),

        "Rows":
            len(powerbi_datasets[dataset_name])

    })


# ------------------------------------------------------------
# EXECUTIVE OUTPUTS
# ------------------------------------------------------------

for output_name in executive_outputs:

    file_path = os.path.join(

        executive_export_path,

        f"{output_name}.csv"

    )

    export_validation.append({

        "Export Group":
            "Executive Output",

        "Object":
            output_name,

        "File Exists":
            os.path.exists(file_path),

        "Rows":
            len(executive_outputs[output_name])

    })


export_validation = pd.DataFrame(
    export_validation
)


all_exports_valid = (
    export_validation["File Exists"]
    .all()
)


display(
    export_validation
)


print("=" * 80)

print(
    f"Total Export Checks : "
    f"{len(export_validation)}"
)

print(
    f"Successful Exports  : "
    f"{export_validation['File Exists'].sum()}"
)

print(
    f"Failed Exports      : "
    f"{(~export_validation['File Exists']).sum()}"
)


if all_exports_valid:

    print(
        "\nFINAL EXPORT STATUS : PASS"
    )

    print(
        "All required project datasets and executive outputs were exported successfully."
    )

else:

    print(
        "\nFINAL EXPORT STATUS : FAIL"
    )

    print(
        "One or more export files require review."
    )
````
